```python
import os
import re
import json

# Directory containing movie/series files
WATCH_FOLDER = "/Users/xiao/Movies/test"

# Log file to store rename history
LOG_FILE = os.path.join(WATCH_FOLDER, "rename_log.json")

# Supported file extensions
MOVIE_EXTENSIONS = [".mp4", ".mkv", ".avi", ".mov"]

# Function to clean and standardize movie/series names
def clean_movie_name(filename):
    # Predefined patterns to remove noise
    patterns = [
        r'(1080p|720p|4K|HDR|HD|BluRay|BRRip|WEB-DL|WEBRip)',  # Resolution
        r'(x264|x265|H264|H265|HEVC|AAC|DTS|DD5\.1|FLAC)',  # Encoding
        r'(\[.*?\]|\(.*?\))',  # Remove content inside brackets
        r'(-|_|\.|\s)+',  # Normalize separators
        r'(\b中字\b|\b国粤双语\b|\b外挂字幕\b|\b特效字幕\b)',  # Remove subtitle tags
        r'(免费在线观看|高清版|完整版|国际版|爱壹帆|腾讯视频|爱奇艺|优酷|哔哩哔哩|B站|Netflix|HBO|Disney\+)',  # Remove streaming service tags
    ]

    # 1. Match filenames that already contain season numbers
    match = re.search(r'(.+?)第(\d+)季[-_\s]*(\d+)', filename)
    if match:
        show_name = match.group(1).strip()
        season = f"SE{int(match.group(2)):02d}"  # Convert to SE02 format
        episode = f"ep{int(match.group(3)):02d}"  # Convert to ep06 format
        return f"{show_name} {season}{episode}".strip()

    # 2. Detect series without an explicit season number (assume SE01)
    match_episode_only = re.search(r'(.+?)[-_\s](\d{2})\b', filename)
    if match_episode_only:
        show_name = match_episode_only.group(1).strip()
        episode = f"ep{int(match_episode_only.group(2)):02d}"  # Convert to ep04

        # If the name suggests a TV series, assume SE01
        if not is_probable_movie(show_name):
            return f"{show_name} SE01{episode}".strip()
        else:
            return f"{show_name} {episode}".strip()  # If it's a movie, keep only episode number

    # 3. Clean generic movie names
    cleaned_name = filename
    for pattern in patterns:
        cleaned_name = re.sub(pattern, " ", cleaned_name, flags=re.IGNORECASE)

    return cleaned_name.strip()

# Function to determine if the file is likely a movie
def is_probable_movie(name):
    """
    - TV shows typically have "-XX" or "EXX" in the filename.
    - Movies usually do not have episode-like patterns.
    - Example:
        - "万物既伟大又渺小-04" → Likely a TV show
        - "流浪地球2" → Likely a movie
    """
    if re.search(r'[-_\s](\d{2})$', name):  # Matches -04, E05 at the end
        return False  # Considered a TV show
    return True  # Considered a movie

# Function to rename files and track changes
def rename_movies_in_folder(folder_path):
    rename_log = {}

    # Load existing rename log to avoid reprocessing files
    if os.path.exists(LOG_FILE):
        with open(LOG_FILE, "r", encoding="utf-8") as f:
            rename_log = json.load(f)

    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)

        # Process only supported video files
        if os.path.isfile(file_path) and any(filename.endswith(ext) for ext in MOVIE_EXTENSIONS):
            file_ext = os.path.splitext(filename)[1]
            original_name = os.path.splitext(filename)[0]  # Remove extension

            # Skip files that have already been renamed
            if filename in rename_log.values():
                continue

            clean_name = clean_movie_name(original_name)  # Process filename
            new_filename = f"{clean_name}{file_ext}"  # Construct new filename
            new_path = os.path.join(folder_path, new_filename)

            # Rename only if there is an actual change in the filename
            if original_name != clean_name:
                os.rename(file_path, new_path)
                rename_log[new_filename] = filename  # Log original name
                print(f"Renamed: {filename} -> {new_filename}")

    # Save rename history
    with open(LOG_FILE, "w", encoding="utf-8") as f:
        json.dump(rename_log, f, indent=4, ensure_ascii=False)

    print("✅ Batch renaming completed!")
    print(f"📄 Rename log saved at {LOG_FILE}")

# Run batch renaming
if __name__ == "__main__":
    rename_movies_in_folder(WATCH_FOLDER)
```
